```python
import undetected_chromedriver as uc
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.keys import Keys
import time
import pandas as pd

driver = uc.Chrome()

# 검색까지 완료
driver.get("https://land.naver.com/")
time.sleep(3)

search_input = driver.find_element(By.ID, "queryInputHeader")
search_input.send_keys("영등포 아트자이")
time.sleep(1)
search_input.send_keys(Keys.RETURN)
time.sleep(5)

print(f"현재 URL: {driver.current_url}")

# 1. 동일매물 묶기 클릭
def click_merge_checkbox():
    """동일매물 묶기 체크박스 클릭"""
    selectors = [
        # 가능한 여러 선택자 시도
        "//input[@id='sameAddressGroup']",
        "//label[contains(text(), '동일매물')]",
        "//span[contains(text(), '동일매물')]",
        "input[id*='same']",
        "input[id*='merge']",
        "label[for*='same']",
        "#sameAddressGroup",
        "input[type='checkbox'][id*='Address']",
    ]
    
    for selector in selectors:
        try:
            if selector.startswith("//"):
                element = driver.find_element(By.XPATH, selector)
            else:
                element = driver.find_element(By.CSS_SELECTOR, selector)
            
            # 체크박스가 이미 체크되어 있는지 확인
            if element.tag_name == 'input':
                if not element.is_selected():
                    element.click()
                    print(f"✓ 동일매물 묶기 활성화 (선택자: {selector})")
                else:
                    print(f"✓ 동일매물 묶기 이미 활성화됨")
            else:
                # label인 경우 클릭
                element.click()
                print(f"✓ 동일매물 묶기 클릭 (선택자: {selector})")
            
            time.sleep(2)
            return True
            
        except Exception as e:
            continue
    
    print("⚠ 동일매물 묶기 버튼을 찾을 수 없습니다. 수동으로 클릭해주세요.")
    print("10초 대기 중...")
    time.sleep(10)
    return False

# 동일매물 묶기 실행
click_merge_checkbox()

# 2. 스크롤하여 모든 매물 로드
def load_all_items():
    """스크롤로 모든 매물 로드"""
    print("\n매물 로딩 시작...")
    
    # 왼쪽 리스트 영역 찾기
    try:
        list_area = driver.find_element(By.CSS_SELECTOR, "div.list_contents, div[class*='list']")
        is_container_scroll = True
        print("✓ 리스트 컨테이너 발견")
    except:
        list_area = None
        is_container_scroll = False
        print("⚠ 리스트 컨테이너 미발견 - 전체 페이지 스크롤 사용")
    
    prev_count = 0
    stable_count = 0
    
    for i in range(100):  # 최대 100회
        # 스크롤
        if is_container_scroll and list_area:
            driver.execute_script("arguments[0].scrollTop = arguments[0].scrollHeight", list_area)
        else:
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        
        time.sleep(1.5)
        
        # 현재 매물 수 확인
        try:
            items = driver.find_elements(By.CSS_SELECTOR, "div[class*='item']")
            current_count = len(items)
            
            print(f"스크롤 {i+1:3d}회 | 매물 수: {current_count:3d}개", end='')
            
            if current_count == prev_count:
                stable_count += 1
                print(f" (변화없음 {stable_count}/5)")
                if stable_count >= 5:
                    print("\n✓ 로딩 완료!")
                    break
            else:
                stable_count = 0
                print(f" (+{current_count - prev_count})")
            
            prev_count = current_count
            
        except Exception as e:
            print(f"\r스크롤 {i+1}회 완료")
    
    time.sleep(2)
    return prev_count

# 모든 매물 로드
total_items = load_all_items()
print(f"\n최종 매물 수: {total_items}개\n")

# 3. 데이터 추출
items = driver.find_elements(By.CSS_SELECTOR, "div[class*='item']")
data_list = []

print("데이터 추출 시작...\n")
for idx, item in enumerate(items, 1):
    try:
        text = item.text.strip()
        if text:
            data_list.append({
                '순번': idx,
                '매물정보': text
            })
            if idx % 20 == 0:
                print(f"{idx}/{len(items)} 추출 완료...")
    except:
        continue

# 4. 저장
df = pd.DataFrame(data_list)
df.to_csv('영등포아트자이_매물.csv', index=False, encoding='utf-8-sig')

print(f"\n✓ {len(df)}개 매물 저장 완료!")
print(f"  - CSV: 영등포아트자이_매물.csv")
print(f"  - Excel: 영등포아트자이_매물.xlsx")
print(f"\n데이터 미리보기:")
print(df.head())
```
